```python
from google.colab import userdata
```

```python
from google import genai
```

```python
from google.genai import types
```

```python
client = genai.Client(
    api_key=userdata.get('gemini_ai')
)
```

```python
MODEL = "gemini-3.8-flash"
```

```python
def ask(prompt, system="You are a helpful tutor.", history=None):

    contents = (history or []) + [prompt]

    r = client.models.generate_content(
        model=MODEL,
        contents=contents,
        config=types.GenerateContentConfig(
            system_instruction=system,
            max_output_tokens=1024
        )
    )

    return r.text

print(ask("Explain LLMs in 3 lines for school students"))
```
